# Ideal Lowpass Filter for 2-D Signals: Continuous and Discrete Signals (image)

# I – Bộ lọc thông thấp lý tưởng cho tín hiệu liên tục 2 chiều

Khi làm việc với một bộ lọc thông thấp lý tưởng hình chữ nhật kích thước $P \times Q$ trong tín hiệu 2D (ví dụ như ảnh), nguyên lý vẫn tương tự như bộ lọc thông thấp hình tròn, nhưng hình dạng vùng thông (pass-band) thay đổi. Thay vì dùng bán kính để xác định tần số cắt, ta xác định các tần số cắt theo hai trục $u$ và $v$, tạo thành vùng elip hoặc hình chữ nhật trong miền tần số.

---

## A – Bộ lọc thông thấp hình chữ nhật cho tín hiệu liên tục

### 1. Biến đổi Fourier

Với tín hiệu 2D $f(x, y)$, biến đổi Fourier được tính như sau:

$$
F(u, v) = \int_{-\infty}^{\infty} \int_{-\infty}^{\infty} f(x, y) e^{-i2\pi(ux + vy)} \, dx \, dy
$$

---

### 2. Hàm lọc

Bộ lọc thông thấp hình chữ nhật lý tưởng có đáp ứng tần số:

$$
H(u, v) = 
\begin{cases} 
1 & \text{nếu } |u| \le D_u \text{ và } |v| \le D_v \\
0 & \text{ngược lại}
\end{cases}
$$

Trong đó $D_u$ và $D_v$ là tần số cắt theo hai trục $u$ và $v$.  
Điều này tạo thành một hình chữ nhật trong miền tần số, tâm tại gốc.

---

### 3. Tín hiệu sau lọc

Trong miền tần số:

$$
G(u, v) = H(u, v) \cdot F(u, v)
$$

---

### 4. Biến đổi ngược

Quay lại miền không gian:

$$
g(x, y) = \int_{-\infty}^{\infty} \int_{-\infty}^{\infty} G(u, v) e^{i2\pi(ux + vy)} \, du \, dv
$$

---

# B – Bộ lọc thông thấp hình tròn 2D (liên tục)

Trong xử lý ảnh, khi chuyển ảnh sang miền tần số, phổ thu được chứa các thành phần tần số đại diện cho mức độ chi tiết khác nhau của ảnh. Bộ lọc thông thấp hình tròn được dùng để làm suy giảm các thành phần tần số cao (thường là nhiễu hoặc chi tiết sắc nét), đồng thời giữ lại thành phần tần số thấp (cấu trúc tổng thể của ảnh).

---

## Công thức toán học

### 1. Biến đổi Fourier của ảnh

$$
F(u, v) = \int \int f(x, y) e^{-j 2\pi (ux + vy)} \, dx \, dy
$$

---

### 2. Tọa độ tần số

Thông thường thành phần tần số 0 được dịch về tâm phổ để dễ quan sát và xử lý.

---

### 3. Hàm lọc hình tròn

$$
H(u, v) =
\begin{cases}
1 & \text{nếu } \sqrt{(u-u_0)^2 + (v-v_0)^2} \le D_0 \\
0 & \text{ngược lại}
\end{cases}
$$

Trong đó $D_0$ là tần số cắt và $(u_0, v_0)$ là tâm phổ.

---

### 4. Áp dụng bộ lọc

$$
G(u, v) = H(u, v) \cdot F(u, v)
$$

---

### 5. Biến đổi ngược

$$
g(x, y) = \int \int G(u, v) e^{j 2\pi (ux + vy)} \, du \, dv
$$

---

## Ứng dụng

- Giảm nhiễu
- Làm mịn ảnh
- Nhấn mạnh cấu trúc lớn

Tuy nhiên bộ lọc lý tưởng có ranh giới cắt sắc nét nên có thể gây hiện tượng ringing. Do đó các bộ lọc như Gaussian hoặc Butterworth thường được ưu tiên hơn.

---

# B – Bộ lọc thông thấp hình tròn lý tưởng cho tín hiệu rời rạc

Trong thực tế, ảnh số là rời rạc và hữu hạn kích thước $P \times Q$.

## 1. DFT 2D của ảnh

$$
F(u,v) = \sum_{x=0}^{P-1} \sum_{y=0}^{Q-1} f(x,y) e^{-i2\pi \left(\frac{ux}{P} + \frac{vy}{Q}\right)}
$$

---

## 2. Hàm lọc

$$
H(u,v) =
\begin{cases}
1 & \text{nếu } 
\sqrt{\left(\frac{u-P/2}{P/2}\right)^2 + \left(\frac{v-Q/2}{Q/2}\right)^2}
\le \frac{D_0}{\min(P,Q)} \\
0 & \text{ngược lại}
\end{cases}
$$

---

## 3. Áp dụng lọc

$$
G(u,v) = H(u,v) \cdot F(u,v)
$$

---

## 4. Biến đổi ngược

$$
g(x,y) = \frac{1}{PQ}
\sum_{u=0}^{P-1}
\sum_{v=0}^{Q-1}
G(u,v) e^{i2\pi \left(\frac{ux}{P} + \frac{vy}{Q}\right)}
$$

---

# C – Bộ lọc thông thấp hình chữ nhật cho tín hiệu rời rạc

Hàm lọc:

$$
H(u,v)=
\begin{cases}
1 & \text{nếu } 
|u-\frac{P}{2}| \le \frac{D_0}{2}
\text{ và }
|v-\frac{Q}{2}| \le \frac{D_0}{2} \\
0 & \text{ngược lại}
\end{cases}
$$

---

# So sánh bộ lọc hình chữ nhật và hình tròn

| Tiêu chí | Hình chữ nhật | Hình tròn |
|-----------|---------------|------------|
| Ưu điểm | Dễ cài đặt, chọn lọc theo hướng | Lọc đẳng hướng, ít méo hình học |
| Nhược điểm | Méo hình học, anisotropy, ringing | Khó kiểm soát hướng, vẫn có ringing |
| Ứng dụng | Khi cần ưu tiên hướng cụ thể | Làm mịn tổng quát |

---

# Kết luận

Việc chọn bộ lọc phụ thuộc vào yêu cầu bài toán:

- Hình tròn: phù hợp cho làm mịn tổng quát.
- Hình chữ nhật: phù hợp khi cần xử lý tần số theo hướng cụ thể.

Cả hai đều có thể gây ringing do biên cắt sắc nét, vì vậy trong thực tế thường dùng bộ lọc có chuyển tiếp mượt hơn.

#### D - CODE IMPLEMENTATION FOR 2D CIRCULAR LOWPASS FILTER

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.fftpack import fftshift, ifftshift, fft2, ifft2
from skimage import data, img_as_float

# Load image
image = img_as_float(data.camera())

# FFT of the image
F = fftshift(fft2(image))

# Generate the filter mask
rows, cols = image.shape
crow, ccol = rows // 2, cols // 2
D0 = 50
H = np.zeros((rows, cols), dtype=np.float32)
for u in range(rows):
    for v in range(cols):
        D = np.sqrt((u - crow) ** 2 + (v - ccol) ** 2)
        H[u, v] = 1 if D <= D0 else 0

# Apply filter in frequency domain
G = H * F

# Inverse FFT to get the filtered image
g = np.real(ifft2(ifftshift(G)))

# Display the original and filtered images
plt.figure(figsize=(12, 6))
plt.subplot(121), plt.imshow(image, cmap="gray"), plt.title("Original Image")
plt.subplot(122), plt.imshow(g, cmap="gray"), plt.title("Filtered Image (Ideal LPF)")
plt.show()
```

#### E - CODE IMPLEMENTATION FOR RECTANGULAR LOWPASS FILTER

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.fftpack import fftshift, ifftshift, fft2, ifft2
from skimage import data, img_as_float

# Load image
image = img_as_float(data.camera())

# FFT of the image
F = fftshift(fft2(image))

# Generate the filter mask for a rectangular lowpass filter
rows, cols = image.shape
crow, ccol = rows // 2, cols // 2
D0 = 50  # Half-width of the rectangular filter

# Initialize the filter mask
H = np.zeros((rows, cols), dtype=np.float32)

# Apply the rectangular condition
for u in range(rows):
    for v in range(cols):
        if abs(u - crow) <= D0 and abs(v - ccol) <= D0:
            H[u, v] = 1  # Inside the rectangle

# Apply filter in frequency domain
G = H * F

# Inverse FFT to get the filtered image
g = np.real(ifft2(ifftshift(G)))

# Display the original and filtered images
plt.figure(figsize=(12, 6))
plt.subplot(121), plt.imshow(image, cmap='gray'), plt.title('Original Image')
plt.subplot(122), plt.imshow(g, cmap='gray'), plt.title('Filtered Image (Rectangular LPF)')
plt.show()
```

### F - CODE IMPLEMENTATION FOR CUT-OFF FREQUENCY FOR BOTH RECTANGULAR AND CIRCULAR LP FILTER

#### Setting the Cutoff Frequency

**Cutoff Frequency $D_0$:**
The choice of $D_0$ is crucial in determining the behavior of the ILPF. The cutoff frequency defines the boundary between the retained low frequencies and the eliminated high frequencies. The process to set $D_0$ involves considering several factors:

1. **Desired Level of Smoothing:**
   - Lower $D_0$ values result in more pronounced smoothing, as more high-frequency content is removed. This is useful for reducing noise but at the risk of losing significant image details.
   - Higher $D_0$ values preserve more detail but may retain more noise.

2. **Nature of the Image/Signal:**
   - Images with finer details may require a higher $D_0$ to preserve these details.
   - For signals or images predominantly composed of lower frequencies, a lower $D_0$ might suffice.

3. **Frequency Analysis:**
   - Analyzing the frequency content of the image or signal can guide the setting of $D_0$. For example, if frequency analysis reveals that significant information content drops beyond a certain frequency, this point can inform an appropriate $D_0$.

#### Practical Considerations and Limitations

While ILPFs are conceptually simple and widely taught for educational purposes, their practical use is limited due to the side effects caused by their abrupt cutoff, such as ringing artifacts due to the Gibbs phenomenon. More advanced filters, such as Butterworth or Gaussian lowpass filters, offer smoother transitions and are often used in practical applications to avoid such artifacts.

In summary, when implementing an ILPF, the choice of $D_0$ should balance between adequate noise reduction and preservation of important image details, tailored to the specific requirements of the application.


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.fftpack import fftshift, ifftshift, fft2, ifft2
from skimage import data, img_as_float

# Load image
image = img_as_float(data.camera())

# FFT of the image
F = fftshift(fft2(image))

# Generate the filter masks
rows, cols = image.shape
crow, ccol = rows // 2 , cols // 2
D0 = 50  # Cutoff radius

# Circular filter mask
H_circular = np.zeros((rows, cols), dtype=np.float32)
for u in range(rows):
    for v in range(cols):
        D = np.sqrt((u - crow)**2 + (v - ccol)**2)
        H_circular[u, v] = 1 if D <= D0 else 0

# Rectangular filter mask
H_rectangular = np.zeros((rows, cols), dtype=np.float32)
r_half = c_half = D0  # Rectangular dimensions based on D0
H_rectangular[crow-r_half:crow+r_half, ccol-c_half:ccol+c_half] = 1

# Apply filters in frequency domain
G_circular = H_circular * F
G_rectangular = H_rectangular * F

# Inverse FFT to get the filtered images
g_circular = np.real(ifft2(ifftshift(G_circular)))
g_rectangular = np.real(ifft2(ifftshift(G_rectangular)))

# Display the original and filtered images
plt.figure(figsize=(18, 6))
plt.subplot(131), plt.imshow(image, cmap='gray'), plt.title('Original Image')
plt.subplot(132), plt.imshow(g_circular, cmap='gray'), plt.title('Circular LPF')
plt.subplot(133), plt.imshow(g_rectangular, cmap='gray'), plt.title('Rectangular LPF')
plt.show()
```

### G - Visualization for Circular Lowpass Filter


```python
import numpy as np
import matplotlib.pyplot as plt
from numpy.fft import ifftshift, ifft2

# Set the size of the image and filter
P, Q = 512, 512  # Example dimensions, you can change these as needed
D0 = 50  # Radius of the circular ILPF

# Create meshgrid
x = np.linspace(-Q/2, Q/2 - 1, Q)
y = np.linspace(-P/2, P/2 - 1, P)
X, Y = np.meshgrid(x, y)
D = np.sqrt(X**2 + Y**2)  # Distance from center

# Create the ideal low-pass filter
H = np.zeros((P, Q))
H[D <= D0] = 1

# Visualize the Frequency Domain Representation
plt.figure(figsize=(12, 6))
plt.subplot(121)
plt.imshow(H, extent=(-Q/2, Q/2 - 1, -P/2, P/2 - 1), cmap='gray')
plt.title("Circular ILPF in Frequency Domain")
plt.colorbar()

# Compute the inverse Fourier Transform to get the spatial representation
h = ifftshift(ifft2(ifftshift(H)))
h_magnitude = np.abs(h)

# Visualize the Time Domain Representation
plt.subplot(122)
plt.imshow(h_magnitude, extent=(-Q/2, Q/2 - 1, -P/2, P/2 - 1), cmap='gray')
plt.title("Spatial (Time) Domain Representation")
plt.colorbar()

plt.show()

```

### Visualization for Ideal Rectangular Lowpass Filter


```python
import numpy as np
import matplotlib.pyplot as plt
from numpy.fft import ifftshift, ifft2

# Set the size of the image and filter
P, Q = 512, 512  # Example dimensions, you can change these as needed
D0 = 50  # Width/Height of the rectangular low-pass filter

# Create meshgrid
x = np.linspace(-Q/2, Q/2 - 1, Q)
y = np.linspace(-P/2, P/2 - 1, P)
X, Y = np.meshgrid(x, y)

# Create the rectangular low-pass filter
H = np.zeros((P, Q))
H[(abs(X) <= D0/2) & (abs(Y) <= D0/2)] = 1  # Rectangular condition

# Visualize the Frequency Domain Representation
plt.figure(figsize=(12, 6))
plt.subplot(121)
plt.imshow(H, extent=(-Q/2, Q/2 - 1, -P/2, P/2 - 1), cmap='gray')
plt.title("Rectangular ILPF in Frequency Domain")
plt.colorbar()

# Compute the inverse Fourier Transform to get the spatial representation
h = ifftshift(ifft2(ifftshift(H)))
h_magnitude = np.abs(h)

# Visualize the Time Domain Representation
plt.subplot(122)
plt.imshow(h_magnitude, extent=(-Q/2, Q/2 - 1, -P/2, P/2 - 1), cmap='gray')
plt.title("Spatial (Time) Domain Representation")
plt.colorbar()

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from numpy.fft import fftshift, ifftshift, fft2, ifft2
import ipywidgets as widgets
from IPython.display import display

# Set the size of the image and filter
P, Q = 512, 512
D0_init = 50

# Create meshgrid
x = np.linspace(-Q/2, Q/2 - 1, Q)
y = np.linspace(-P/2, P/2 - 1, P)
X, Y = np.meshgrid(x, y)

# Initialize plot
fig, ax = plt.subplots(2, 2, figsize=(12, 12))
plt.subplots_adjust(hspace=0.3, wspace=0.3)
im1 = ax[0, 0].imshow(np.zeros((P, Q)), extent=(-Q/2, Q/2 - 1, -P/2, P/2 - 1), cmap='gray')
ax[0, 0].set_title("Circular Filter in Frequency Domain")
im2 = ax[1, 0].imshow(np.zeros((P, Q)), extent=(-Q/2, Q/2 - 1, -P/2, P/2 - 1), cmap='gray')
ax[1, 0].set_title("Circular Filter Spatial Domain")
im3 = ax[0, 1].imshow(np.zeros((P, Q)), extent=(-Q/2, Q/2 - 1, -P/2, P/2 - 1), cmap='gray')
ax[0, 1].set_title("Rectangular Filter in Frequency Domain")
im4 = ax[1, 1].imshow(np.zeros((P, Q)), extent=(-Q/2, Q/2 - 1, -P/2, P/2 - 1), cmap='gray')
ax[1, 1].set_title("Rectangular Filter Spatial Domain")

# Widgets for control
cutoff_slider = widgets.IntSlider(value=D0_init, min=0, max=250, step=1, description='Cutoff Frequency:')
zoom_slider = widgets.FloatSlider(value=1, min=1, max=10, step=0.1, description='Zoom:')
pan_slider = widgets.FloatSlider(value=0, min=-0.5, max=0.5, step=0.01, description='Pan:')

def update(change):
    D0 = cutoff_slider.value
    zoom = zoom_slider.value
    pan = pan_slider.value

    # Circular Filter
    D = np.sqrt(X**2 + Y**2)
    H_circular = (D <= D0).astype(int)
    h_circular = ifftshift(ifft2(ifftshift(H_circular)))
    
    # Rectangular Filter
    H_rectangular = np.zeros((P, Q))
    H_rectangular[int(P/2-D0):int(P/2+D0), int(Q/2-D0):int(Q/2+D0)] = 1
    h_rectangular = ifftshift(ifft2(ifftshift(H_rectangular)))
    
    extent = np.array([-Q/2, Q/2 - 1, -P/2, P/2 - 1]) * zoom + pan * Q
    
    # Update plots
    im1.set_data(H_circular)
    im1.set_extent(extent)
    im1.set_cmap('gray')
    im1.set_clim(0, 1)
    
    im2.set_data(np.abs(h_circular))
    im2.set_extent(extent)
    im2.set_cmap('gray')
    im2.set_clim(0, np.abs(h_circular).max())

    im3.set_data(H_rectangular)
    im3.set_extent(extent)
    im3.set_cmap('gray')
    im3.set_clim(0, 1)
    
    im4.set_data(np.abs(h_rectangular))
    im4.set_extent(extent)
    im4.set_cmap('gray')
    im4.set_clim(0, np.abs(h_rectangular).max())
    
    fig.canvas.draw_idle()

cutoff_slider.observe(update, names='value')
zoom_slider.observe(update, names='value')
pan_slider.observe(update, names='value')

display(cutoff_slider, zoom_slider, pan_slider)
update(None)  # Initial update call to set the plots correctly
```
